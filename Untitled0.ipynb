```python
from transformers import pipeline
```

```python
classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)
```

```python
text = "I love to learn about Transformers. They are amazing to learn, but I don't like it to learn. I will use ChatGPT for the code."

result = classifier(text)

print(result)
```

```python
sentences = [
    "Date with her was fantastic.",
    "She was looking amazing.",
    "I am feeling so good with her.",
    "Overall day was good with her."
]

results = classifier(sentences)

for sentence, result in zip(sentences, results):
    print(sentence, result)
```
